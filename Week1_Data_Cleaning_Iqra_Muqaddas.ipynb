# Importing Necessary Libraries

```python
import pandas as pd
import numpy as np
```

# Loading the Dataset

```python
df = pd.read_csv("/content/messy_applicants.csv")
df.head()
```

 # Understanding Data Before Cleaning

```python

df_shape = df.shape

print("=="*60)
print("shape of df" , df_shape  )
print("=="*60)

print(df.info())
```

**The dataset has 1000 rows and 8 columns where Applicant Name has 50 missing values, Email has 150 missing values, University has 350 missing values, Status has 150 missing values. There are 0 duplicate records, and Application Date is currently stored as text (object type), not as an
actual date."**

```python
missing_values = df.isnull().sum() # exact count of missing values per column
duplicates = df.duplicated().sum()  # how many exact duplicate rows exist
print(" Missing values:\n" , missing_values)

print("="*60)

print("Duplicates:", duplicates)
```

#Handle Missing Values

**I dropped rows with missing Applicant Name because a nameless application isn't usable for outreach.**

```python
df.dropna(subset=['Applicant_Name', 'Email'], how='all', inplace=True)
```

```python
df.isnull().sum()
```

**I filled the missing Email records with "not_provided@example.com" because we have phone numbers and names through which we can outreach the applicants**



```python
df['Email'] = df['Email'].fillna("not_provided@example.com")
```

# Standardize Text Fields

**Checking leading and trailing whitespaces in Applicant Column**

```python
for name in df['Applicant_Name']:
  if name != name.strip():
    print(repr(name))
```

**Removing leading and trailing whitespaces from Applicant Names**

```python
df['Applicant_Name'] = df['Applicant_Name'].str.strip()
```

**Checking the inconsistent names in the University column**

```python
df['University'].value_counts()
```

**I filled missing University with "University of Engineering and Technology" instead of dropping because the inconsistent patterns shows data of the same university**

```python
df["University"] = df['University'].fillna("University of Engineering and Technology")
```

**Checking Inconsistencies in the Status Column**

```python
df['Status'].value_counts()
```

**I filled the missing Status values with Under Review, it likely means the application hasn't been processed yet**

```python
df["Status"] = df['Status'].fillna("Under Review")
```

```python
df.isnull().sum()
```

**Checking Inconsistencies in the Domain Applied Column**

```python
df['Domain_Applied'].value_counts()
```

**Replacing inconsistencies with uniform domain values**

```python
df['Domain_Applied'] = df['Domain_Applied'].replace({
    'web dev': 'Web Development',
    'Web development': 'Web Development',
    'WEB DEV': 'Web Development',
    "data science": "Data Science",
    "DATA SCIENCE": "Data Science",
    "Data science": "Data Science",
    "AI": "Artificial Intelligence",
    "artificial intelligence": "Artificial Intelligence",
    "ARTIFICIAL INTELLIGENCE": "Artificial Intelligence"
})
```

**Checking that the formatting is consistent now in Domain Applied Column**

```python
df['Domain_Applied'].unique()
```

**Replacing inconsistencies with uniform values in both University and Status column**

```python
df["University"] = df['University'].replace({
    "UET" : "University of Engineering and Technology",
    "U.E.T": "University of Engineering and Technology",
    "uet": "University of Engineering and Technology",
    "Uet lahore": "University of Engineering and Technology"
})

df['Status'] = df["Status"].replace({
    "under review": "Under Review",
    "UNDER REVIEW": "Under Review",
    "Under review": "Under Review",
    "selected": "Selected",
    "SELECTED": "Selected",
    "sElEcTeD": "Selected",
    "rejected": "Rejected",
    "REJECTED": "Rejected",
    "rejected": "Rejected"
})
```

**Checking that the formatting is consistent now in University Column**

```python
df['University'].unique()
```

```python
df['Email'].head(10)
```

**Removing the whitespsaces from Email Column and Formatting the mixed cases to lower cases**

```python
df['Email'] = df['Email'].str.strip().str.lower()
```

```python
df['Application_Date'].head()
```

**Converting the Application Date column which as mixed date formatting to consistent format and changing its datatype to datetime**

```python
df['Application_Date'] = pd.to_datetime(
    df['Application_Date'],
    format='mixed',
    dayfirst=True,
    errors='coerce'
)
df['Application_Date']
```

```python
df['Phone']
```

**Removing all the whitespaces and other characters from phone numbers and storing it as string datatype**

```python
df['Phone'] = df['Phone'].astype('string').str.strip()

df['Phone'] = df['Phone'].str.replace(
    r'\D', '', regex=True
)
df['Phone']
```

# Verifying fixes worked

```python
df.info()
```

```python
df.to_csv("cleaned_applicants.csv", index=False)
```
